```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Example: assume your data is already in a pandas DataFrame
data_dint = pd.read_csv("C:/Users/atanoh/Documents/2025/stage/C03_Torres_2005/CAP/data_dint.csv")

# Drop NA values for EvolDiasUCI
df = data_dint.dropna(subset=["EvolDiasUCI"])

# --- p1: Scatterplot of IdPac vs EvolDiasUCI ---
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df, x="IdPac", y="EvolDiasUCI", color="steelblue")
# plt.title("Length of stay in ICU", fontsize=14, weight="bold")
plt.xlabel("Id Patients")
plt.ylabel("Number of days")
plt.xticks([])  # Hide x-axis labels like in ggplot
sns.despine()
plt.show()
```

```python
# --- p2: Histogram (barplot) of EvolDiasUCI counts ---
plt.figure(figsize=(8, 5))
sns.countplot(data=df, x="vf_days", color="steelblue")
# plt.title("Number of Patients by length of stay in the ICU", fontsize=14, weight="bold")
plt.xlabel("Days")
plt.ylabel("Number of Patients")
sns.despine()
plt.show()
```
