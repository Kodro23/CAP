```python
from scipy.stats import gamma

a = 0.1
b = 0.1 # c’est le *rate* (lambda)
scale = 1 / b  # attention: scipy utilise le paramètre scale = 1/rate

# quantiles
q025 = gamma.ppf(0.025, a, scale=scale)
q975 = gamma.ppf(0.975, a, scale=scale)
print(q025, q975)
```

```python
from scipy.stats import beta

# Beta parameters
a = 0.05
b = 0.001

# Quantiles
q025 = beta.ppf(0.025, a, b)
q975 = beta.ppf(0.975, a, b)

print(q025, q975)
```
